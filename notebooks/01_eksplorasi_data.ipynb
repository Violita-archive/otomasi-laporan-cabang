## 4. Wilayah 3 - Mataraman

Kalau dibaca langsung, header-nya kacau. Dibaca dulu tanpa header untuk lihat bentuk aslinya.

```python
cek = pd.read_excel(RAW + "Laporan Wilayah III Mataraman.xlsx", header=None)
cek.head(7)
```

Ada 3 baris judul dan 1 baris kosong di atas. Header tabel ada di baris ke-5
(indeks 4), jadi dibaca ulang dengan `header=4`.

```python
wil3 = pd.read_excel(RAW + "Laporan Wilayah III Mataraman.xlsx", header=4)
print(wil3.shape)
wil3.head()
```

```python
wil3.dtypes
```

Angka ditulis sebagai teks dengan awalan `Rp` dan koma sebagai pemisah ribuan
(`Rp 1,115,601.88`). Formatnya berbeda dengan Wilayah 2 yang pakai titik.

Bulan ditulis `01/2024`, kode cabang berupa angka biasa (`13`), dan ada kolom `No` yang tidak diperlukan.

```python
# Variasi penulisan nama cabang
wil3["Nama Cabang"].value_counts().sort_index()
```

Nama cabang tidak konsisten. Satu cabang bisa ditulis tiga cara, misalnya
`Madiun`, `Kota Madiun`, dan `MADIUN`. Karena itu penggabungan nanti lebih
aman pakai kode cabang, bukan nama.

```python
# Cek total DPK, angka diubah dulu dari teks
def rp_ke_angka(kolom):
    return kolom.str.replace("Rp ", "").str.replace(",", "").astype(float)

total = rp_ke_angka(wil3["Total DPK"])
jumlah = rp_ke_angka(wil3["Giro"]) + rp_ke_angka(wil3["Tabungan"]) + rp_ke_angka(wil3["Deposito"])
selisih = (total - jumlah).round(2)

wil3[selisih != 0][["Bulan", "Kode", "Nama Cabang", "Total DPK"]].assign(selisih=selisih)
```

```python
print("Sel kosong:", wil3.isna().sum().sum())
print("Baris duplikat:", wil3.duplicated().sum())
```

### Yang perlu dilakukan saat cleaning

1. Menyamakan nama kolom di semua file wilayah.
2. Mengubah semua periode ke format `YYYY-MM`.
3. Menyamakan kode cabang ke format `KC001`, lalu nama, kab/kota, dan wilayah diambil dari master.
4. Mengubah angka yang masih berupa teks menjadi angka.
5. Menghapus baris duplikat.
6. Memperbaiki angka yang satuannya salah.
7. Menentukan cara menangani sel kosong dan bulan yang hilang.
8. Mengecek ulang total DPK. Total dihitung ulang dari giro + tabungan + deposito, bukan memakai kolom total dari file.
9. Mengubah target ke format memanjang lalu menggabungkannya dengan data realisasi.
